```python
bad_tags = ['visual_separator', 'mua_signature', 'patch', 'personal_signature', 'quotation_marker', 'inline_headers', 'section_heading']
bad_tags = [f'<gmane_tag_{t}>' for t in bad_tags]

def highlight_non_tag_text(text):
    # Define the pattern to match tags in square brackets
    tag_pattern = r'\[[^\[\]]+\]'

    # Split the text based on the tag pattern, keeping the tags
    segments = re.split(f'({tag_pattern})', text)

    # Wrap non-tag segments with HTML light gray highlighting
    highlighted_text = ''
    for i, segment in enumerate(segments):
        if re.match(tag_pattern, segment):  # Tag segments
            segment = segment.replace('raw_code', 'code').replace('log_data', 'log')
            highlighted_text += f'<span style="background-color: #edebeb;color: #999999;">{segment[1:-1]}</span>'
        else:  # Non-tag segments
            highlighted_text += f'<span style="background-color: #e8baba;">{segment}</span>'

    return highlighted_text

def process_text(text):
    #print(text)
    for tag in bad_tags:
        text = text.replace(tag, '')
    text = text.replace('gmane_tag_', '').replace('<','[').replace('>',']<br>').replace('\n','<br>')
    text = highlight_non_tag_text(text)
    return text

def clean_text(text):
    #print(text)
    for tag in bad_tags:
        text = text.replace(tag, '')
    text = text.replace('gmane_tag_', '')
    return text
```

```python
df['text'] = ['%s <br> <hr size="2"> %s'%(f'<span style="background-color: #e8baba;">{s}</span>',process_text(it)) for s,it in zip(df['subject'], df['message_body'])]
df = df.drop_duplicates('text')
df['id'] = [i for i in range(len(df))]
```

```python
df['cleaned_text'] = ['%s \n %s'%(s,clean_text(it)) for s,it in zip(df['subject'], df['message_body'])]
```
