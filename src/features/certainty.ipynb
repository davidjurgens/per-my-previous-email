```python
import os
```

```python
import sys
import torch
sys.path.append("/shared/2/projects/jiaxin/hf_models/")
from inference import Estimator
```

```python
def get_certainty(texts):
    os.environ["CUDA_DEVICE_ORDER"]="PCI_BUS_ID"   
    os.environ["CUDA_VISIBLE_DEVICES"]="0"
    model = Estimator(tokenizer='pedropei/sentence-level-certainty', model='pedropei/sentence-level-certainty',num_labels=1,cuda = False)
    return model.predict(texts)
```

```python
import pandas as pd
```

```python
file_list = os.listdir("/shared/4/projects/research-jam-2024/working-dir/monthly-posts-cleaned/")
for i in range(len(file_list) - 1, -1, -1):
  if "2009" in file_list[i]:
      print(file_list[i])
      file = os.path.join("/shared/4/projects/research-jam-2024/working-dir/monthly-posts-cleaned/", file_list[i])
      df = pd.read_csv(file, usecols=["message_id", "message_body_clean"], sep='\t')
      # df = df.iloc[0:1000]
      df = df.dropna()
      
      ids = df.message_id.to_list()
      content = df['message_body_clean'].to_list()
      pred = get_certainty(content)
      
      df_certainty = pd.DataFrame({"id":ids, "content":content, "pred":pred})
      new_filename = "/shared/4/projects/research-jam-2024/working-dir/certainty/" + file_list[i] 
      df_certainty.to_csv(new_filename, sep='\t', index=False)
      del df_certainty, df, content, ids, pred
```
