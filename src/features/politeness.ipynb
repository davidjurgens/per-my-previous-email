```python
import os
os.environ["CUDA_DEVICE_ORDER"]="PCI_BUS_ID"   
os.environ["CUDA_VISIBLE_DEVICES"]="0,1,2"
import pandas as pd
import torch
import sys
sys.path.append("/shared/2/projects/jiaxin/hf_models/politeness")
from politeness_inference import PolitenessEstimator
```

```python
def get_politeness(texts):
    model = PolitenessEstimator(cuda = True)
    return model.predict(texts, max_length=128)
```

```python
file_list = os.listdir("/shared/4/projects/research-jam-2024/working-dir/monthly-posts-cleaned/")
for i in range(len(file_list) - 1, -1, -1):
  if "2009" in file_list[i] and ("-04" in file_list[i] or "-06" in file_list[i] or "-08" in file_list[i] or "-09" in file_list[i] or "-11" in file_list[i] or "-12" in file_list[i]):
      print(file_list[i])
      file = os.path.join("/shared/4/projects/research-jam-2024/working-dir/monthly-posts-cleaned/", file_list[i])
      df = pd.read_csv(file, usecols=["message_id", "message_body_clean"], sep='\t')    
      df = df.dropna()
    
      ids = df.message_id.to_list()
      content = df['message_body_clean'].to_list()
      pred = get_politeness(content)
      
      
      df_polite = pd.DataFrame({"id":ids, "content":content, "pred":pred})
      new_filename = "/shared/4/projects/research-jam-2024/working-dir/politeness/" + file_list[i] 
      df_polite.to_csv(new_filename, sep='\t', index=False)
      del df_polite, df, content, ids, pred
```
