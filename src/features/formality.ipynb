```python
def get_formality(texts):
    model = Estimator(tokenizer='s-nlp/roberta-base-formality-ranker', model='s-nlp/roberta-base-formality-ranker',num_labels=2,cuda = True)
    return model.predict(texts, max_length=128)
```

```python
import pandas as pd
import torch
```

```python
file_list = os.listdir("/shared/4/projects/research-jam-2024/working-dir/monthly-posts-cleaned/")
for i in range(len(file_list) - 1, -1, -1):
  if "2009" in file_list[i] and ("-03" not in file_list[i] and "-07" not in file_list[i] and "-10" not in file_list[i]):
      print(file_list[i])
      file = os.path.join("/shared/4/projects/research-jam-2024/working-dir/monthly-posts-cleaned/", file_list[i])
      df = pd.read_csv(file, usecols=["message_id", "message_body_clean"], sep='\t')    
      df = df.dropna()
    
      ids = df.message_id.to_list()
      content = df['message_body_clean'].to_list()
      pred = get_formality(content)
      
      df_formality = pd.DataFrame({"id":ids, "content":content, "pred":pred})
      new_filename = "/shared/4/projects/research-jam-2024/working-dir/formality/" + file_list[i] 
      df_formality.to_csv(new_filename, sep='\t', index=False)
      del df_formality, df, content, ids, pred
```
