```python
def escape_gmane_tags(msg):
    # Escape <gmane_tag_*> elements by replacing < and > with their HTML entities
    return re.sub(r'(<gmane_tag_[^>]+>)', lambda x: x.group(0).replace('<', '&lt;').replace('>', '&gt;'), msg)


def clean_up_message_for_html(msg):
    msg = re.sub(r'\n+', '\n', msg)  # Consolidate multiple new lines into a single new line

    # Convert tabs and sequences of spaces to HTML non-breaking spaces
    msg = msg.replace('\t', '&nbsp;' * 4)

    # Escape gmane tags to ensure they are not treated as HTML by the browser
    msg = escape_gmane_tags(msg)

    # Wrap text in paragraphs
    paragraphs = msg.split('\n')
    msg = ''.join(f'<p>{paragraph}</p>' if paragraph.strip() else '' for paragraph in paragraphs)

    return msg

Write code to apply clean_up_message_for_html to dataframe column message_body and call it message_body_html
```

```python
samples = []
for i, message_body in enumerate(df['message_body']):
    samples.append(message_body)
    if i == 5:
        break
```

```python
for sample in samples:
    print(sample)
    print("\n\n")
    display(HTML(clean_up_message_for_html(sample)))
    print("\n\n")
```
