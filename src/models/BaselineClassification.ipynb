```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.multioutput import MultiOutputClassifier
from sklearn.dummy import DummyClassifier
from sklearn.pipeline import Pipeline
```

```python
df = pd.read_csv('../../annotation/email_understanding/annotation_output/full/df4model.csv')
df = df.dropna()
```

```python
columns = list(df.columns)

label_columns = [x for x in columns if 'label' in x]


texts = df['cleaned_text']
labels = df[label_columns].values
```

```python
X_train, X_test, y_train, y_test = train_test_split(texts, labels, test_size=0.2, random_state=42)
```

```python
pipeline = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('dummy_clf', MultiOutputClassifier(DummyClassifier(strategy='uniform')))
])

pipeline.fit(X_train, y_train)

```

```python
y_pred = pipeline.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, hamming_loss

# Calculating metrics for each label
precision_per_label = precision_score(y_test, y_pred, average=None, zero_division=0) 
recall_per_label = recall_score(y_test, y_pred, average=None, zero_division=0)  
f1_per_label = f1_score(y_test, y_pred, average=None, zero_division=0)  
# Printing the metrics for each label
label_names = label_columns  # Assuming y_test is a DataFrame; adjust as necessary
print("Scores for each label:")
for label, prec, rec, f1 in zip(label_names, precision_per_label, recall_per_label, f1_per_label):
    print(f"{label}: Precision = {prec:.4f}, Recall = {rec:.4f}, F1 Score = {f1:.4f}")

# Note: Hamming Loss is calculated globally, not per label, so it remains unchanged
print(f"Hamming Loss: {hamming_loss(y_test, y_pred):.4f}")
```
