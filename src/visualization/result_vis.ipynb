```python
import altair as alt


import pandas as pd
data = {
    '3': ['dummy', 0.299, 'micro_f1'],
    '4': ['dummy', 0.257, 'macro_f1'],
    '9': ['log_reg', 0.518, 'micro_f1'],
    '10': ['log_reg', 0.169, 'macro_f1'],
    '1': ['roberta_b', 0.790, 'micro_f1'],
    '2':  ['roberta_b', 0.540, 'macro_f1'],
    '5': ['liama_2_m', 0.410, 'micro_f1'],
    '6': ['liama_2_m', 0.336, 'macro_f1'],
    '7': ['liama_2_s', 0.430, 'micro_f1'],
    '8': ['liama_2_s', 0.351, 'macro_f1'],
}
source = pd.DataFrame.from_dict(data, orient='index')
source.columns = ['model', 'value', 'metrics']
source


alt.Chart(source).mark_bar().encode(
    x=alt.X('metrics:O'),
    y=alt.Y('sum(value):Q'),
    color=alt.Color('metrics:N'),
    column=alt.Color('model:N')
).properties(
    width=80,
    height=400
)
```
